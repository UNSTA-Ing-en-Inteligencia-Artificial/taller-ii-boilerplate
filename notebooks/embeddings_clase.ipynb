## Embeddings pre-entrenados (español)
FastText entrenado sobre Common Crawl y Wikipedia, 300 dimensiones. Se debe descargar desde:
`https://dl.fbaipublicfiles.com/fasttext/vectors-crawl/cc.es.300.vec.gz`

Cargamos solo las 200.000 palabras más frecuentes para que entre en memoria.

```python
from gensim.models import KeyedVectors

es = KeyedVectors.load_word2vec_format("../bin/cc.es.300.vec.gz", limit=200_000)
print(es.most_similar("rey", topn=5))
print(es.most_similar(positive=["rey", "mujer"], negative=["hombre"], topn=5))
print(es.most_similar(positive=["París", "Italia"], negative=["Francia"], topn=5))
```

```python
pares_en = [("king", "queen"), ("car", "banana"), ("doctor", "nurse")]
pares_es = [("rey", "reina"), ("coche", "plátano"), ("médico", "enfermera")]

for (a, b), (c, d) in zip(pares_en, pares_es):
    print(f"{a}-{b}: {glove.similarity(a, b):.2f}   |   {c}-{d}: {es.similarity(c, d):.2f}")
```

## Transformers

```python
import numpy as np

def coseno(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
```

```python
import torch
from transformers import AutoTokenizer, AutoModel

nombre = "bert-base-multilingual-cased"
tokenizador = AutoTokenizer.from_pretrained(nombre)
bert = AutoModel.from_pretrained(nombre).eval()

def vector(oracion, palabra):
    enc = tokenizador(oracion, return_tensors="pt")
    ids = enc["input_ids"][0].tolist()
    pid = tokenizador.convert_tokens_to_ids(palabra)
    assert pid != tokenizador.unk_token_id and pid in ids, f"'{palabra}' no es un único token"
    with torch.no_grad():
        h = bert(**enc).last_hidden_state[0]
    return h[ids.index(pid)].numpy()

plaza   = vector("Me senté en el banco de la plaza.", "banco")
dinero1 = vector("Fui al banco a sacar dinero.", "banco")
dinero2 = vector("El banco aprobó mi préstamo.", "banco")

print("plaza  vs dinero1:", round(float(coseno(plaza, dinero1)), 3))
print("dinero1 vs dinero2:", round(float(coseno(dinero1, dinero2)), 3))
```
